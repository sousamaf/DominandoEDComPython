# Estrutura para Indexação de Eventos por Código

```python
# Definição da classe Evento
class Evento:
    def __init__(self, codigo, titulo, palestrante, tipo, prioridade):
        self.codigo = codigo
        self.titulo = titulo
        self.palestrante = palestrante
        self.tipo = tipo
        self.prioridade = prioridade

    def __str__(self):
        return f"Código: {self.codigo}, Título: {self.titulo}, Palestrante: {self.palestrante}, Tipo: {self.tipo}, Prioridade: {self.prioridade}"

# Definição do nó da ABB com objeto Evento
class Node:
    def __init__(self, evento):
        self.evento = evento
        self.esquerda = None
        self.direita = None

# Estrutura da Árvore Binária de Busca
class ABBEventos:
    def __init__(self):
        self.raiz = None

    def inserir_evento(self, evento):
        self.raiz = self._inserir(self.raiz, evento)

    def _inserir(self, raiz, evento):
        if raiz is None:
            return Node(evento)
        elif evento.codigo < raiz.evento.codigo:
            raiz.esquerda = self._inserir(raiz.esquerda, evento)
        else:
            raiz.direita = self._inserir(raiz.direita, evento)
        return raiz

    def buscar_evento(self, codigo):
        return self._buscar(self.raiz, codigo)

    def _buscar(self, raiz, codigo):
        if raiz is None:
            return None
        if raiz.evento.codigo == codigo:
            return raiz.evento
        elif codigo < raiz.evento.codigo:
            return self._buscar(raiz.esquerda, codigo)
        else:
            return self._buscar(raiz.direita, codigo)

    def percurso_em_ordem(self):
        print("Percurso em ordem (códigos):")
        self._percurso_em_ordem(self.raiz)
        print()

    def _percurso_em_ordem(self, raiz):
        if raiz:
            self._percurso_em_ordem(raiz.esquerda)
            print(raiz.evento.codigo, end=' ')
            self._percurso_em_ordem(raiz.direita)
```

## Demonstração da Solução


```python
# Instanciando a árvore
arvore = ABBEventos()

# Criando e inserindo eventos fictícios
eventos = [
    Evento(30, "Congresso de Tecnologia", "Prof. Graça", "Congresso", "Alta"),
    Evento(15, "Workshop de IA", "Dr. Fábio", "Workshop", "Média"),
    Evento(45, "Palestra sobre Dados", "Dra. Carla", "Palestra", "Alta"),
    Evento(10, "Minicurso de Python", "Prof. Daniel", "Minicurso", "Baixa"),
    Evento(25, "Painel de Inovação", "Prof. Elisa", "Painel", "Média"),
]

for e in eventos:
    arvore.inserir_evento(e)

# Exibindo percurso em ordem
arvore.percurso_em_ordem()

# Buscando um evento pelo código
codigo_busca = 25
evento_encontrado = arvore.buscar_evento(codigo_busca)

print(f"\n\nResultado da busca pelo código {codigo_busca}:")
if evento_encontrado:
    print(evento_encontrado)
else:
    print("Evento não encontrado.")
```
