# Busca Recursiva de Evento por Código

```python
# 1. Definindo a estrutura do nó de evento
class EventoNode:
    def __init__(self, codigo, titulo, tipo, prioridade):
        self.codigo = codigo
        self.titulo = titulo
        self.tipo = tipo
        self.prioridade = prioridade
        self.next = None

# 2. Criando a lista encadeada de eventos
class ListaEventos:
    def __init__(self):
        self.head = None

    def inserir_evento(self, codigo, titulo, tipo, prioridade):
        novo_evento = EventoNode(codigo, titulo, tipo, prioridade)
        if self.head is None:
            self.head = novo_evento
        else:
            atual = self.head
            while atual.next:
                atual = atual.next
            atual.next = novo_evento

    def exibir_lista(self):
        atual = self.head
        while atual:
            print(f"{atual.codigo}: {atual.titulo} ({atual.tipo} - {atual.prioridade})")
            atual = atual.next
```

```python
# 3. Função recursiva para buscar um evento por código
def buscar_evento_por_codigo(node, codigo):
    if node is None:
        return "Evento não encontrado."
    if node.codigo == codigo:
        return f"Evento encontrado: {node.titulo}"
    return buscar_evento_por_codigo(node.next, codigo)
```

```python
# 4. Criando a lista e inserindo três eventos
lista = ListaEventos()
lista.inserir_evento("EVT101", "Congresso Nacional de Pesquisa", "congresso", "alta")
lista.inserir_evento("EVT102", "Workshop de Inovação", "workshop", "média")
lista.inserir_evento("EVT103", "Minicurso de Python", "minicurso", "baixa")

# Exibindo a lista
lista.exibir_lista()


# 5. Executando buscas
print(buscar_evento_por_codigo(lista.head, "EVT102"))  # Esperado: Workshop de Inovação
print(buscar_evento_por_codigo(lista.head, "EVT999"))  # Esperado: Evento não encontrado.
```
