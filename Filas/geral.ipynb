```python
import time
import random
from IPython.display import clear_output
from fila_atendimento import FilaAtendimento

# ----------------------------
# Função para imprimir o estado atual da fila
# ----------------------------
def print_fila(fila):
    atual = fila.cabeca
    clientes = []
    while atual:
        clientes.append(f"[Cliente {atual.valor}]")
        atual = atual.proximo
    print("[Fila atual]: " + " -> ".join(clientes) if clientes else "[Fila atual]: <vazia>")

# ----------------------------
# Simulação simples
# ----------------------------
fila = FilaAtendimento()

# Função para simular o processo: chegada (enqueue) e atendimento (dequeue)
def simula_fila():
    # Simula a chegada de quatro clientes:
    clientes = [1, 2, 3, 4]
    for cliente in clientes:
        tempo_atendimento = random.randint(5, 10)
        fila.enqueue(cliente, tempo_atendimento)
        clear_output(wait=True)
        print(f"Cliente {cliente} chegou (tempo de atendimento: {tempo_atendimento}s)")
        print_fila(fila)
        time.sleep(3)  # pausa para visualização

    # Agora simula o atendimento de um cliente por vez enquanto a fila não estiver vazia
    while True:
        try:
            valor, tempo_atendimento = fila.dequeue()
            clear_output(wait=True)
            print(f"Atendendo Cliente {valor} (tempo: {tempo_atendimento}s)")
            print_fila(fila)
            time.sleep(3)  # pausa para visualização
        except Exception as e:
            clear_output(wait=True)
            print("Fila esvaziada.")
            break

# Executa a simulação
simula_fila()
```
