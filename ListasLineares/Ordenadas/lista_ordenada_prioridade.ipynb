# Lista Ordenada de Eventos por Prioridade

```python
# Classe Node adaptada para representar um evento com prioridade textuall
class Node:
    def __init__(self, descricao, prioridade):
        self.descricao = descricao
        self.prioridade = prioridade  # alta, média, baixa
        self.next = None

# Classe Lista Ordenada por Prioridade
class ListaOrdenadaPorPrioridade:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    # Mapeamento de prioridade para valor comparável
    def prioridade_para_valor(self, prioridade):
        pesos = {"alta": 1, "média": 2, "baixa": 3}
        return pesos.get(prioridade, 99)  # padrão para prioridade inválida

    # Inserção ordenada por prioridade (menor valor = maior prioridade)
    def insert(self, descricao, prioridade):
        novo = Node(descricao, prioridade)
        peso_novo = self.prioridade_para_valor(prioridade)

        if self.head is None or self.prioridade_para_valor(self.head.prioridade) > peso_novo:
            novo.next = self.head
            self.head = novo
        else:
            atual = self.head
            while atual.next and self.prioridade_para_valor(atual.next.prioridade) <= peso_novo:
                atual = atual.next
            novo.next = atual.next
            atual.next = novo

    def display(self):
        if self.is_empty():
            print("A agenda está vazia.")
        else:
            atual = self.head
            print("Agenda de Eventos (ordenada por prioridade):")
            while atual:
                print(f"- {atual.descricao} ({atual.prioridade})")
                atual = atual.next
```

```python
# Criando a lista ordenada de eventos
agenda = ListaOrdenadaPorPrioridade()

# Inserindo eventos
agenda.insert("Congresso Nacional de Pesquisa", "alta")
agenda.insert("Workshop de Inovação", "média")
agenda.insert("Minicurso de Introdução à Programação", "baixa")

# Exibindo a agenda ordenada por prioridade
agenda.display()
```
